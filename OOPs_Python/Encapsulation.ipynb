Inheritancce


```python
#Reusing attributes and from a parent (Base class)

class Employee:
    start_time = "10 pm"
    end_time = "2 am"

    def changeEndTime(self, new_end_time):
        self.end_time = new_end_time


class Teacher(Employee):
    def __init__(self, name):
        self.name = name

class Adminstaff(Employee):
    def __init__(self, role):
        self.role = role


t1 = Teacher("Anish")
print(t1.name, t1.start_time, t1.end_time)
sj = t1.changeEndTime("12 am")
print(t1.end_time)

a1 = Adminstaff("Manager")
print(a1.role, a1.start_time, a1.end_time)


        
```

Types of Inheritance 

1 . Single Level Inheritance

```python
class School:
    def __init__(self, School_name):
        self.School_name = School_name

class Teacher(School):
    def __init__(self, name, School_name):
        super().__init__(School_name)
        self.name = name 


t1 = Teacher("Anish", "PBCOE")
print(f"Name of the School is {t1.School_name} & name of teacher is {t1.name}")

```

2 . Multi Level Inheritance 

```python
class School:
    def __init__(self, School_name):
        self.School_name = School_name

class Teacher(School):
    def __init__(self, name, School_name):
        super().__init__(School_name)
        self.name = name 

class Student(Teacher):
    def __init__(self, student_name, School_name, name ):
        super().__init__(name, School_name)
        self.student_name = student_name
        

s1 = Student("Aniket Gawali","Anish Sir", "PBCOE")
print(f"The name of student is : {s1.student_name} \n His name of School is : {s1.School_name} \n His Class Teacher name is: {s1.name}")

t1 = Teacher("Anish Sir", "PBCOE")
print(f"Name of the School is {t1.School_name} & name of teacher is {t1.name}")

```

3. Multiple Inheritance 

```python
class Teacher:
    def __init__(self,salary):
        self.salary = salary

class Student:
    def __init__(self,gpa):
        self.gpa = gpa

class TA(Teacher, Student):
    def __init__(self, salary, gpa, name):
        super().__init__(salary)
        Student.__init__(self, gpa)
        self.name = name

ta1 = TA(10000, 9.8, "Aniket")
print(ta1.salary)
```
